# Notebook de Manipulação de Dados com pandas

Este notebook apresenta operações básicas com pandas: criar, transformar, salvar e ler dados tabulares.

## 1. Importar Bibliotecas Essenciais

Importe pandas e NumPy, usados para manipulação e análise de dados.

```python
import numpy as np
import pandas as pd
```

## 2. Criar e Exibir um DataFrame Simples

Crie um DataFrame a partir de um dicionário e confira as primeiras linhas.

```python
dados = {
    "produto": ["Caderno", "Caneta", "Mochila"],
    "preco": [12.50, 3.00, 89.90],
    "quantidade": [4, 10, 2],
}

df = pd.DataFrame(dados)
df
```

## 3. Adicionar Novas Colunas ao DataFrame

Calcule o valor total de cada linha multiplicando o preço pela quantidade.

```python
df["valor_total"] = df["preco"] * df["quantidade"]
df
```

## 4. Salvar o DataFrame em um Arquivo CSV

Salve o DataFrame sem incluir o índice como uma coluna adicional.

```python
caminho_csv = "dados_produtos.csv"
df.to_csv(caminho_csv, index=False)
print(f"Arquivo salvo em: {caminho_csv}")
```

## 5. Ler Dados de um Arquivo CSV

Leia o CSV salvo em um novo DataFrame e exiba os dados carregados.

```python
df_lido = pd.read_csv(caminho_csv)
df_lido
```
